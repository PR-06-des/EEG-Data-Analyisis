# EEG Data Exploration

Explore the ds003061 EEG recording before preprocessing or ERP analysis.

```python
from pathlib import Path
import mne
from mne_bids import BIDSPath, read_raw_bids

from src.download import download_subject_data

PROJECT_DIR = Path.cwd()
DATASET_DIR = PROJECT_DIR / 'ds003061'

if not DATASET_DIR.exists():
    download_subject_data(
        dataset_id='ds003061',
        subject_id='sub-001',
        target_dir=str(DATASET_DIR),
    )

print(f'Dataset path: {DATASET_DIR}')
```

```python
bids_path = BIDSPath(
    subject='001',
    task='P300',
    run='1',
    datatype='eeg',
    root=DATASET_DIR,
)

raw = read_raw_bids(bids_path=bids_path, verbose=False)
raw.load_data()

print(raw)
print(f'Channels: {len(raw.ch_names)}')
print(f"Sampling rate: {raw.info['sfreq']} Hz")
print(f'Total duration: {raw.times[-1]:.2f} seconds')
```

```python
print('Channel names:')
print(raw.ch_names)

print('\nChannel types:')
print(raw.get_channel_types())
```

```python
raw.plot(
    n_channels=20,
    duration=20,
    scalings='auto',
    title='Raw EEG signal',
)
```

```python
events, event_id = mne.events_from_annotations(raw)

print('Event IDs:')
print(event_id)
print(f'Total events: {len(events)}')
print(events[:10])
```

```python
raw.compute_psd(fmax=50).plot(
    average=True,
    picks='eeg',
    spatial_colors=False,
    show=True,
)
```

```python
filtered = raw.copy().filter(
    l_freq=1.0,
    h_freq=30.0,
    fir_design='firwin',
)

filtered.plot(
    n_channels=20,
    duration=20,
    scalings='auto',
    title='Band-pass filtered EEG signal',
)
```

## Exploration notes

Use the plots and printed metadata above to inspect channel quality, recording duration, frequency content, and available event annotations.
